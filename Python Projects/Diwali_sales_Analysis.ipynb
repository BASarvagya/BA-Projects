# Importing Libraries & Data Cleaning

```python
import pandas as pd
import numpy as np
import seaborn as srn
import matplotlib as plt
```

```python
df = pd.read_csv(r'C:\Users\jimmy\OneDrive\Desktop\Python\Diwali_Sales_Analysis\Diwali Sales Data.csv', encoding = 'unicode_escape' )
```

```python
df.drop(['Status', 'unnamed1'],axis=1, inplace = True)
```

```python
df.dropna(inplace = True)
df.info()
```

```python
df['Amount'] = df['Amount'].astype(int)
```

```python
df.head()
```

# Exploratory Data Analysis

```python
# Orders by Gender 
Gender_orders = df.groupby(['Gender'], as_index= False) ['Orders'].sum()
srn.barplot( x='Gender',y='Orders', data= Gender_orders)
```

```python
# Amount by Gender
amount_gender = df.groupby(['Gender'], as_index= False) ['Amount'].sum().sort_values(by='Amount', ascending = False)
no = srn.barplot( x='Gender',y='Amount', data= amount_gender)
for bar in no.containers:
    no.bar_label(bar, fmt = '%0f')
```

```python
# Orders by Age Group
age_group_orders = df.groupby(['Age Group'], as_index = False) ['Orders'].sum().sort_values(by='Orders', ascending = False)
srn.barplot( x = 'Age Group', y = 'Orders', data = age_group_orders)
```

```python
# Amount by Age Group
age_group_amount = df.groupby(['Age Group'], as_index=False) ['Amount'].sum().sort_values('Amount', ascending = False)
fox = srn.barplot( x = 'Age Group', y = 'Amount', data = age_group_amount)
for bars in fox.containers:
    fox.bar_label(bars, fmt='%.0f')
```
